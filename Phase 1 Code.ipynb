# Data Cleaning


# Missing Values

```python
missing = pd.DataFrame({
    "count": df.isna().sum(),
    "percent": (df.isna().mean() * 100).round(2)
})
missing[missing["count"] > 0].sort_values("count", ascending=False)
```

```python
df = df.drop(columns="deck")

df["age"] = df["age"].fillna(
    df.groupby(["pclass", "sex"])["age"].transform("median")
)

df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode()[0])

df.isna().sum()
```

# Outlier Detection

```python
num_cols = ["age", "fare", "sibsp", "parch"]
fig, axes = plt.subplots(4, 1, figsize=(16, 16))
for ax, col in zip(axes, num_cols):
    sns.boxplot(y=df[col], ax=ax)
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

```python
def iqr_outliers(s):
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return lower, upper, ((s < lower) | (s > upper)).sum()

print("Column names are capitalized")
for col in num_cols:
    lo, hi, n = iqr_outliers(df[col])
    print(f"{col.capitalize()}: bounds=({lo:.2f}, {hi:.2f}), outliers={n}")
```

# Outlier Treatment Techniques

```python
# Option A: keep as is  # For Ages and Sibsp (Sibling and Spouses abroad ) we can since its not an error

# Option B: cap at the upper IQR bound (winsorizing) # we cap fare since first class is a thing and not usually bought
lo, hi, _ = iqr_outliers(df['fare'])
df['fare_capped'] = df['fare'].clip(upper=hi)


# Option C: log transform

# Option D: remove #data set is small we cant lose much of it
```

# Univariate Analysis

```python
num_cols = ["age", "fare_capped", "sibsp", "parch"]
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, col in zip(axes.ravel(), num_cols):
    sns.histplot(df[col], kde=True, ax=ax)
    ax.axvline(df[col].mean(), color="red", linestyle="--", label="mean")
    ax.axvline(df[col].median(), color="green", linestyle="-", label="median")
    ax.set_title(col)
    ax.legend()
plt.tight_layout()
plt.show()

df[num_cols].agg(["mean", "median", "std", "skew"]).round(2)

# Categorical: counts
cat_cols = ["survived", "pclass", "sex", "embarked", "alone"]
fig, axes = plt.subplots(1, 5, figsize=(20, 4))
for ax, col in zip(axes, cat_cols):
    sns.countplot(x=col, data=df, ax=ax)
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

## Summary

- The dataset has 891 passengers and 15 columns. The target is `survived`: about 62% died and 38% survived.
- `deck` was dropped because most of its values were missing. Missing `age` values were filled with the median, and the 2 missing `embarked` values were filled with the most common port.
- 107 duplicate rows were found but kept, because there is no passenger ID to prove they are the same person.
- `fare` had extreme high values, so it was capped (`fare_capped`) instead of removed. The other outliers were realistic values and were kept.
- Sex had the biggest effect on survival: women survived far more than men.
- Passenger class also mattered: 1st class survived much more than 3rd class.
- Fare and class are strongly related, since higher classes paid more.
- Age had a smaller effect on survival.
